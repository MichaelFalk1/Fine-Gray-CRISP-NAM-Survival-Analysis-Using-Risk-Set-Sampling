```python
import pandas as pd
import numpy as np

# pip install ucimlrepo
from ucimlrepo import fetch_ucirepo

import kagglehub
```

```python

# fetch dataset 
support2 = fetch_ucirepo(id=880) 
  
# data (as pandas dataframes) 
X = support2.data.features 
y = support2.data.targets 
  
# metadata 
# print(support2.metadata) 
  
# variable information 
# print(support2.variables) 



df_support2 = pd.concat([X, y], axis=1)
df_support2.head(50)


```

```python
df_support2["glucose"].unique()
```

```python
df_support2.info()
```

```python



df_support2.columns = df_support2.columns.str.strip().str.lower().str.replace('.', '_', regex=False)


# A tiny number of rows (1 to 2) are missing vital tracking stats like 'meanbp' or 'hrt'.
# It is safest to drop these specific rows.
critical_cols = ['scoma', 'sps', 'aps', 'surv2m', 'surv6m', 'meanbp', 'hrt', 'resp', 'temp', 'sod']
df_support2 = df_support2.dropna(subset=critical_cols)


# This loops through columns with lots of blanks (like 'glucose', 'edu', 'charges') 
# and fills them with the middle value so your models do not crash.
num_cols_with_nas = [
    'edu', 'charges', 'totcst', 'totmcst', 'avtisst', 'prg2m', 'prg6m', 
    'dnrday', 'wblc', 'pafi', 'alb', 'bili', 'crea', 'ph', 'glucose', 
    'bun', 'urine', 'adlp', 'adls'
]
for col in num_cols_with_nas:
    median_value = df_support2[col].median()
    df_support2[col] = df_support2[col].fillna(median_value)

#Fill Missing Text/Categorical Values (Using Mode)
cat_cols_with_nas = ['income', 'race', 'dnr', 'sfdm2']
for col in cat_cols_with_nas:
    most_common = df_support2[col].mode()[0]
    df_support2[col] = df_support2[col].fillna(most_common)


# Cleans up any messy spacing or mixed casing in text columns
text_cols = ['sex', 'dzgroup', 'dzclass', 'income', 'race', 'ca', 'dnr', 'sfdm2']
for col in text_cols:
    df_support2[col] = df_support2[col].astype(str).str.strip().str.lower()


print("--- Cleaning Complete ---")
print(f"Remaining rows: {df_support2.shape[0]}")
print(f"Total missing values left: {df_support2.isna().sum().sum()}")
```

```python
df_framingham = pd.read_csv('framingham.csv')

df_framingham.head()
```

```python

import pandas as pd
import numpy as np

# 1. Clean Column Names
# Converts "currentSmoker" or "TenYearCHD" to "current_smoker" and "ten_year_chd"
df_framingham.columns = (
    df_framingham.columns
    .str.replace(r'(?<!^)(?=[A-Z])', '_', regex=True)
    .str.lower()
    .str.strip()
)

df_framingham = df_framingham.rename(columns={"b_m_i": "bmi"})
df_framingham = df_framingham.rename(columns={"b_p_meds": "bp_meds"})

# 3. Smart Imputation: Cigarettes Per Day
# If someone is a smoker but cigs_per_day is blank, fill it with the median for smokers.
# If they are NOT a smoker, fill it with 0.
is_smoker = df_framingham['current_smoker'] == 1
smoker_median_cigs = df_framingham.loc[is_smoker, 'cigs_per_day'].median()

df_framingham.loc[is_smoker, 'cigs_per_day'] = df_framingham.loc[is_smoker, 'cigs_per_day'].fillna(smoker_median_cigs)
df_framingham.loc[~is_smoker, 'cigs_per_day'] = df_framingham.loc[~is_smoker, 'cigs_per_day'].fillna(0)

# 4. Fill Medical Missing Values (Using Median)
# Fills missing values for cholesterol, BMI, heart rate, and glucose using midpoints.
med_cols_to_fill = ['tot_chol', 'bmi', 'heart_rate', 'glucose']
for col in med_cols_to_fill:
    median_value = df_framingham[col].median()
    df_framingham[col] = df_framingham[col].fillna(median_value)

# 5. Fill Categorical/Binary Missing Values (Using Mode)
# Education and BP Meds are categories, so we fill blanks with the most common choice.
cat_cols_to_fill = ['education', 'bp_meds']
for col in cat_cols_to_fill:
    most_common_value = df_framingham[col].mode()[0]
    df_framingham[col] = df_framingham[col].fillna(most_common_value)

# 6. Fix Data Types
# Columns like education and bp_meds should be whole numbers, not decimals (floats).
df_framingham['education'] = df_framingham['education'].astype(int)
df_framingham['bp_meds'] = df_framingham['bp_meds'].astype(int)
df_framingham['cigs_per_day'] = df_framingham['cigs_per_day'].astype(int)

# Verify your cleaned dataset
print("--- Framingham Cleaning Complete ---")
print(f"Remaining rows: {df_framingham.shape[0]}")
print(f"Total missing values left: {df_framingham.isna().sum().sum()}")
```

```python
df_framingham.to_csv('framingham.csv', index=False)
```
